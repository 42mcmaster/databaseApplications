## Sub-Lesson 02a: Load & Summarize

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')   # lets charts render without a pop-up window
import matplotlib.pyplot as plt

df = pd.read_csv('pizza_orders.csv')
print('rows, columns:', df.shape)
print(df.head(3))
```

```python
# The analyst's first summary
print('Total revenue: $', round(df['line_total'].sum(), 2))
print('Average quantity per order:', round(df['quantity'].mean(), 2))
print()
print(df[['unit_price','quantity','line_total']].describe())
```

### Group and total (the pandas `GROUP BY`)

```python
rev_by_region = df.groupby('region')['line_total'].sum().round(2)
print(rev_by_region)
print()
print('Top region:', rev_by_region.idxmax())
```

## Sub-Lesson 02b: Chart It

Each chart answers a different kind of question.

```python
# BAR: compare a total across categories
ax = rev_by_region.plot(kind='bar', title='Revenue by Region')
ax.set_ylabel('Revenue ($)')
plt.tight_layout(); plt.savefig('chart_bar.png'); plt.close()
print('saved chart_bar.png')
```

```python
# HISTOGRAM: the distribution of one number
ax = df['line_total'].plot(kind='hist', bins=8, title='Distribution of Order Totals')
ax.set_xlabel('line_total ($)')
plt.tight_layout(); plt.savefig('chart_hist.png'); plt.close()
print('saved chart_hist.png')
```

```python
# SCATTER: the relationship between two numbers
ax = df.plot(kind='scatter', x='quantity', y='line_total', title='Quantity vs. Order Total')
plt.tight_layout(); plt.savefig('chart_scatter.png'); plt.close()
print('saved chart_scatter.png')
```
